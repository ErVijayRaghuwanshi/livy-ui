```python
from pyspark.sql import functions as F

df = (
    spark.range(1_000_000)
    .withColumn("value", F.rand(seed=42))
    .withColumn(
        "category",
        F.when(F.col("value") < 0.25, "A")
         .when(F.col("value") < 0.50, "B")
         .when(F.col("value") < 0.75, "C")
         .otherwise("D")
    )
)

df.createOrReplaceTempView("test_large_result")
```

```python
spark.sql("""
    SELECT *
    FROM test_large_result
    LIMIT 10
""").show()
```

```python
spark.sql("SELECT COUNT(*) FROM test_large_result").show()
```

```python
page_size = 1000

page = (
    spark.table("test_large_result")
    .orderBy("id")
    .limit(page_size)
)

pdf = page.toPandas()

pdf.head()
```

```python
page = (
    spark.table("test_large_result")
    .orderBy("id")
    .offset(1000001)
    .limit(1000)
)

pdf = page.toPandas()

pdf.head()
```

```python
page = (
    spark.table("test_large_result")
    .orderBy("id")
    .offset(2000)
    .limit(1000)
)

pdf = page.toPandas()
```
